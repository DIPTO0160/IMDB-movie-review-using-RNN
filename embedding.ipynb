## Embedding of the data


```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
##sentence example
sentence=[
    "the glass of milk",
    "the glass o juice",
    "the cup of tea",
    "I am a good boy",
    "I am a good developer",
    "understands the meaning of the word"
]
```

```python
## define the vocabulary size
voc_size=10000
```

```python
## one hot representation
one_hot_representation=[one_hot(words,voc_size) for words in sentence]
```

```python
one_hot_representation
```

```python
## word embedding representation
from tensorflow.keras.layers import Embedding,Input
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
max_sentence_length=8
embedded_docs=pad_sequences(one_hot_representation,padding="pre",maxlen=max_sentence_length)
```

```python
print(embedded_docs)
```

```python
## feature representation
dim=10
```

```python
model=Sequential()
model.add(Input(shape=(max_sentence_length,)))
model.add(Embedding(voc_size,dim))
model.compile(optimizer="adam",loss="mse")
```

```python
model.summary()
```

```python
model.predict(embedded_docs)[0]
```
